## 1. Forward Euler discretization

Approximate the derivative at time $t$ using a forward difference:

$$
\frac{dC}{dt}\approx
\frac{C(t+\Delta t)-C(t)}{\Delta t}.
$$

Forward Euler evaluates the right-hand side at the start of the step:

$$
\frac{C(t+\Delta t)-C(t)}{\Delta t}
=5-0.25C(t).
$$

Solving for the concentration at the next time step gives

$$
{
C(t+\Delta t)=C(t)+\Delta t[5-0.25C(t)]
}.
$$

Collecting terms,

$$
{
C(t+\Delta t)=(1-0.25\Delta t)C(t)+5\Delta t
}.
$$

For $\Delta t=1$, this becomes

$$
C(t+1)=0.75C(t)+5.
$$

## 2. Characteristic timescale and step-size assessment

For a system of the form

$$
\frac{dC}{dt}=a-bC,
$$

deviations from equilibrium decay exponentially with characteristic
timescale T=1/|Rate|, where "b" is the "slope/rate" of the formula. Therefore,

$$
\boxed{T=\frac{1}{0.25}=4\text{ time units}}.
$$

With $\Delta t=1$,

$$
\frac{\Delta t}{T}=\frac{1}{4}=0.25.
$$

This gives four steps per characteristic timescale, so it is a reasonable
starting choice, although a smaller step may be needed for high accuracy.

The forward Euler stability condition is

$$
|1-0.25\Delta t|<1,
$$

which gives

$$
0<\Delta t<8.
$$

Thus, $\Delta t=1$ is stable. Accuracy should be checked by reducing the
step size and comparing the results.

## 3. Predicted error

In the first-order convergence regime, the global error is proportional
to the step size:

$$
E(\Delta t)\approx K\Delta t.
$$

Therefore,

$$
E_{\mathrm{new}}
\approx E_{\mathrm{old}}
\frac{\Delta t_{\mathrm{new}}}{\Delta t_{\mathrm{old}}}.
$$

Using the given values,

$$
E(1)\approx 0.6\left(\frac{1}{2}\right).
$$

Solution:

$$
\boxed{E(1)\approx 0.3}.
$$

#### Problem 1.2 (5)

A well-mixed holding tank has volume $V = 200\ \text{m}^3$ and a

    $\Delta t = 1\ \text{yr}$ a reasonable starting choice of step size?
    Justify your answer.

    $\Delta t = 1.6\ \text{yr}$. What error would you expect at
    $\Delta t = 0.4\ \text{yr}$?

## 1. Forward Euler discretization

Approximate the derivative using a forward difference:

$$
\frac{dC}{dt}\approx
\frac{C(t+\Delta t)-C(t)}{\Delta t}.
$$

Evaluate all concentration terms on the right-hand side at time $t$:

$$
\frac{C(t+\Delta t)-C(t)}{\Delta t}
=\frac{Q}{V}[C_{\mathrm{in}}-C(t)]-kC(t).
$$

Solving for the next concentration gives

$$
{
C(t+\Delta t)
=C(t)+\Delta t
\left(
\frac{Q}{V}[C_{\mathrm{in}}-C(t)]-kC(t)
\right)
}.
$$

Collecting terms,

$$
{
C(t+\Delta t)
=\left[1-\left(\frac{Q}{V}+k\right)\Delta t\right]C(t)
+\frac{Q}{V}C_{\mathrm{in}}\Delta t
}.
$$

Substituting the numerical values, with $\Delta t$ in years,

$$
{
C(t+\Delta t)=(1-0.55\Delta t)C(t)+0.5\Delta t
}.
$$

For $\Delta t=1\ \mathrm{yr}$, this becomes

$$
C(t+1)=0.45C(t)+0.5.
$$

Here, the additive concentration is $0.5\ \mathrm{mg/L}$.

## 2. Characteristic timescale and step-size assessment

Flushing and decay both remove pollutant, so their rates add:

$$
b=\frac{Q}{V}+k=0.25+0.3=0.55\ \mathrm{yr^{-1}}.
$$

The characteristic timescale is therefore

$$
{
T=\frac{1}{Q/V+k}
=\frac{1}{0.55}
\approx 1.82\ \mathrm{yr}
}.
$$

For $\Delta t=1\ \mathrm{yr}$,

$$
\frac{\Delta t}{T}=\frac{1}{1.82}\approx 0.55.
$$

This gives fewer than two steps per characteristic timescale. This step is
not a good choice for accurately resolving the response, although it can be used for a rough first estimate.

Therefore, a better forward Euler stability condition is

$$
0<\Delta t<\frac{2}{0.55}\approx 3.64\ \mathrm{yr}.
$$

Thus, $\Delta t=1\ \mathrm{yr}$ is stable, but stability alone does not
mean better accuracy. A smaller starting step, such as
$\Delta t=0.1T\approx 0.18\ \mathrm{yr}$, would better resolve the
response. Step-size refinement would be good for better accuracy.

## 3. Predicted error

For convergence,

$$
E_{\mathrm{new}}
\approx E_{\mathrm{old}}
\frac{\Delta t_{\mathrm{new}}}{\Delta t_{\mathrm{old}}}.
$$

Using the given values,

$$
E(0.4)\approx 0.8\left(\frac{0.4}{1.6}\right)
=0.8(0.25).
$$

Therefore,

$$
\boxed{E(0.4)\approx 0.2}.
$$

The predicted error has the same units or error measure as the given error.
